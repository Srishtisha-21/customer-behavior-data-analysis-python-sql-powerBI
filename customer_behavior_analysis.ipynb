## Treating Null Values

### 1.Size

```python
df['Size'].unique()
```

```python
#checking where where null values is present in category
df.groupby('Category')['Size'].unique()
```

```python
#checking where where null values is present in item purchased
df.groupby('Item Purchased')['Size'].unique()
```

```python
df.groupby('Category')['Size'].unique()
```

```python
#treating null values or replacing all values in electronics category
df.loc[df['Category']=='Electronics','Size']="Not Applicable"
```

```python
#treating null values or replacing all values in Accessories category
df.loc[df['Category']=='Accessories','Size']="Free Size"
```

```python
#treating null values or replacing all values in Footwear category
df.loc[df['Category']=='Footwear','Size']="Not Available"
```

```python
#treating null values or replacing all values in Clothing  category
df.loc[(df['Category']=='Clothing') & (df['Size'].isnull()),'Size'] =df[df['Category']=='Clothing']['Size'].mode()[0]
```

```python
df.groupby('Category')['Size'].unique()
```

```python
#checking if we still have any null value in size
df.isnull().sum()
```

## 2.Review Rating


```python
#mean for review ratings
df['Review Rating'].mean()
```

```python
#median for review ratings
df['Review Rating'].median()
```

```python
#mode for review ratings
df['Review Rating'].mode()
```

```python
#checking which item purchase get average rating
df.groupby("Item Purchased")['Review Rating'].mean()
```

```python
#filling null values of Review Ratings with the mean of each Item Purchased
df['Review Rating']=df['Review Rating'].fillna(df.groupby('Item Purchased')['Review Rating'].transform('mean'))
```

```python
#checking if review rating has still any null values
df.isnull().sum()
```

```python
df['Review Rating']=df['Review Rating'].round(2)
```

## 3.Previous Purchased

```python
#filling previous purchases with zero(0)
df['Previous Purchases']=df['Previous Purchases'].fillna(0)
```

```python
df.isnull().sum()
```

## 4.Purchase Amount (USD)

```python
df.groupby("Item Purchased")['Purchase Amount (USD)'].mean()
```

```python
#filling null values of Review Purchase Amount (USD) with the mean of each Item Purchased
df['Purchase Amount (USD)']=df['Purchase Amount (USD)'].fillna(df.groupby('Item Purchased')['Purchase Amount (USD)'].transform('mean'))
```

```python
df.isnull().sum()
```

## Renaming the Header

```python
df.columns
```

```python
df.columns=df.columns.str.replace(" ","_")
df.columns=df.columns.str.lower()

#replacing purchase_amount_(usd) to purchase_amount
df=df.rename(columns={"purchase_amount_(usd)":"purchase_amount"})
```

```python
df.columns
```

```python
#create a column age_group
labels=['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
#create column purchase_frequency_days

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Monthly' : 30,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```
